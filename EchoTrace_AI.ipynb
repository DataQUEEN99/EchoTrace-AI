```python
# ================================================================
# 🧠 ECHOTRACE AI
# AI MEMORY MUTATION & RELIABILITY OBSERVATORY
#
# COMPLETE ONE-CELL GOOGLE COLAB PROJECT
# ================================================================

# ================================================================
# 0. INSTALL
# ================================================================

!pip -q install sentence-transformers networkx plotly seaborn scikit-learn

# ================================================================
# 1. IMPORTS
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx
import plotly.express as px

from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

from IPython.display import display, Markdown

np.random.seed(42)

sns.set_theme(
    style="whitegrid",
    context="notebook"
)

print("=" * 90)
print("                         ECHOTRACE AI")
print("              AI MEMORY & RELIABILITY OBSERVATORY")
print("=" * 90)

# ================================================================
# 2. LOAD SEMANTIC MODEL
# ================================================================

print("\nLoading semantic embedding model...")

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("✓ Embedding model loaded")


# ================================================================
# 3. QUESTION BANK
# ================================================================

questions = [

    {
        "id": "Q001",
        "question": "What is CRISPR?",
        "topic": "Biotechnology",
        "canonical": "CRISPR is a genome editing technology that allows scientists to modify DNA at specific locations.",
        "facts": [
            "genome editing",
            "DNA modification",
            "specific locations"
        ],
        "entities": [
            "CRISPR",
            "DNA"
        ]
    },

    {
        "id": "Q002",
        "question": "What is machine learning?",
        "topic": "Artificial Intelligence",
        "canonical": "Machine learning is a field of artificial intelligence in which algorithms learn patterns from data to make predictions or decisions.",
        "facts": [
            "artificial intelligence",
            "learn patterns",
            "data",
            "predictions"
        ],
        "entities": [
            "machine learning",
            "artificial intelligence"
        ]
    },

    {
        "id": "Q003",
        "question": "What is DNA?",
        "topic": "Biology",
        "canonical": "DNA is a molecule that stores hereditary genetic information in living organisms.",
        "facts": [
            "molecule",
            "genetic information",
            "hereditary"
        ],
        "entities": [
            "DNA",
            "genetic information"
        ]
    },

    {
        "id": "Q004",
        "question": "What is cloud computing?",
        "topic": "Technology",
        "canonical": "Cloud computing provides computing resources such as storage and processing through remote networked infrastructure.",
        "facts": [
            "computing resources",
            "storage",
            "processing",
            "remote infrastructure"
        ],
        "entities": [
            "cloud computing"
        ]
    },

    {
        "id": "Q005",
        "question": "What is Python?",
        "topic": "Programming",
        "canonical": "Python is a high-level general-purpose programming language widely used for software development, automation, data analysis and artificial intelligence.",
        "facts": [
            "programming language",
            "software development",
            "data analysis",
            "artificial intelligence"
        ],
        "entities": [
            "Python"
        ]
    },

    {
        "id": "Q006",
        "question": "What is an API?",
        "topic": "Software",
        "canonical": "An API is an interface that allows different software systems to communicate and exchange data or functionality.",
        "facts": [
            "interface",
            "software systems",
            "communication",
            "data exchange"
        ],
        "entities": [
            "API"
        ]
    },

    {
        "id": "Q007",
        "question": "What is a neural network?",
        "topic": "Deep Learning",
        "canonical": "A neural network is a machine learning model composed of interconnected computational units that learn patterns from data.",
        "facts": [
            "machine learning",
            "interconnected units",
            "patterns",
            "data"
        ],
        "entities": [
            "neural network",
            "machine learning"
        ]
    },

    {
        "id": "Q008",
        "question": "What is an antibiotic?",
        "topic": "Medicine",
        "canonical": "An antibiotic is a substance used to treat bacterial infections by inhibiting bacterial growth or killing bacteria.",
        "facts": [
            "bacterial infections",
            "bacterial growth",
            "bacteria"
        ],
        "entities": [
            "antibiotic",
            "bacteria"
        ]
    },

    {
        "id": "Q009",
        "question": "What is blockchain?",
        "topic": "Technology",
        "canonical": "Blockchain is a distributed digital ledger that records transactions across a network using cryptographic techniques.",
        "facts": [
            "distributed ledger",
            "transactions",
            "network",
            "cryptography"
        ],
        "entities": [
            "blockchain",
            "cryptography"
        ]
    },

    {
        "id": "Q010",
        "question": "What is bioinformatics?",
        "topic": "Bioinformatics",
        "canonical": "Bioinformatics uses computational methods to analyze biological data such as DNA, RNA and protein sequences.",
        "facts": [
            "computational methods",
            "biological data",
            "DNA",
            "RNA",
            "protein sequences"
        ],
        "entities": [
            "bioinformatics",
            "DNA",
            "RNA",
            "proteins"
        ]
    }
]


# ================================================================
# 4. SIMULATED AI RESPONSE GENERATOR
#
# This makes the project reproducible without requiring an API.
# In the production version, replace this with actual LLM calls.
# ================================================================

mutation_types = [

    "stable",
    "paraphrase",
    "fact_mutation",
    "entity_mutation",
    "contradiction",
    "shortening",
    "expansion"
]


def generate_response(question_data, day, model_name):

    base = question_data["canonical"]

    # Model-specific mutation tendency
    if model_name == "Echo-Alpha":
        mutation_probability = 0.08

    elif model_name == "Echo-Beta":
        mutation_probability = 0.15

    else:
        mutation_probability = 0.22

    # Time increases mutation probability
    time_factor = min(
        0.25,
        day / 300
    )

    probability = (
        mutation_probability +
        time_factor
    )

    random_value = np.random.random()

    if random_value > probability:

        return base, "stable"

    mutation = np.random.choice(
        mutation_types[1:]
    )

    # ------------------------------------------------------------
    # Paraphrase
    # ------------------------------------------------------------

    if mutation == "paraphrase":

        if "is a" in base:

            text = base.replace(
                "is a",
                "can be described as a"
            )

        else:

            text = (
                "In simple terms, "
                + base
            )

        return text, "paraphrase"


    # ------------------------------------------------------------
    # Fact mutation
    # ------------------------------------------------------------

    if mutation == "fact_mutation":

        replacements = {

            "genome editing":
                "genome sequencing",

            "DNA modification":
                "RNA analysis",

            "artificial intelligence":
                "database technology",

            "genetic information":
                "metabolic information",

            "computing resources":
                "physical computing devices",

            "distributed digital ledger":
                "centralized database",

            "bacterial infections":
                "viral infections",

            "computational methods":
                "laboratory methods",

            "data analysis":
                "graphic design"

        }

        text = base

        changed = False

        for old, new in replacements.items():

            if old in text:

                text = text.replace(
                    old,
                    new
                )

                changed = True
                break

        if not changed:

            text = (
                base +
                " It is mainly used for unrelated applications."
            )

        return text, "fact_mutation"


    # ------------------------------------------------------------
    # Entity mutation
    # ------------------------------------------------------------

    if mutation == "entity_mutation":

        replacements = {

            "CRISPR":
                "PCR",

            "DNA":
                "RNA",

            "Python":
                "Java",

            "blockchain":
                "traditional database",

            "machine learning":
                "robotics",

            "bioinformatics":
                "biochemistry",

            "antibiotic":
                "antiviral drug"

        }

        text = base

        for old, new in replacements.items():

            if old in text:

                text = text.replace(
                    old,
                    new
                )

                break

        return text, "entity_mutation"


    # ------------------------------------------------------------
    # Contradiction
    # ------------------------------------------------------------

    if mutation == "contradiction":

        contradictions = {

            "CRISPR":
                "CRISPR does not modify DNA; it is primarily a method for measuring blood chemistry.",

            "machine learning":
                "Machine learning does not learn patterns from data and instead depends entirely on manually written rules.",

            "DNA":
                "DNA does not store hereditary information and has no role in biological inheritance.",

            "cloud computing":
                "Cloud computing requires all computing resources to be physically located on the user's own computer.",

            "Python":
                "Python is a hardware description language rather than a general-purpose programming language.",

            "API":
                "An API prevents software systems from communicating with one another.",

            "neural network":
                "Neural networks do not learn patterns from data.",

            "antibiotic":
                "Antibiotics are primarily designed to treat viral infections.",

            "blockchain":
                "Blockchain is a centralized ledger controlled by one database administrator.",

            "bioinformatics":
                "Bioinformatics does not use computational methods to study biological data."

        }

        keyword = question_data["question"].split(
            "?"
        )[0].replace(
            "What is ",
            ""
        )

        for key, value in contradictions.items():

            if key.lower() in keyword.lower():

                return value, "contradiction"

        return (
            "The original explanation is not considered accurate.",
            "contradiction"
        )


    # ------------------------------------------------------------
    # Shortening
    # ------------------------------------------------------------

    if mutation == "shortening":

        words = base.split()

        shortened = " ".join(
            words[:max(8, len(words)//2)]
        )

        return (
            shortened + ".",
            "shortening"
        )


    # ------------------------------------------------------------
    # Expansion
    # ------------------------------------------------------------

    if mutation == "expansion":

        return (
            base +
            " The concept is widely discussed in modern scientific and technological research.",
            "expansion"
        )


    return base, "stable"


# ================================================================
# 5. GENERATE LONGITUDINAL EXPERIMENT
# ================================================================

days = [
    1,
    7,
    14,
    30,
    60,
    90
]

models = [
    "Echo-Alpha",
    "Echo-Beta",
    "Echo-Gamma"
]


records = []

print(
    "\nGenerating longitudinal AI response experiment..."
)

for model in models:

    for q in questions:

        for day in days:

            response, mutation = generate_response(
                q,
                day,
                model
            )

            records.append({

                "Model":
                    model,

                "Question_ID":
                    q["id"],

                "Question":
                    q["question"],

                "Topic":
                    q["topic"],

                "Day":
                    day,

                "Response":
                    response,

                "Mutation_Type":
                    mutation

            })


responses_df = pd.DataFrame(
    records
)


print(
    "✓ Generated responses:",
    len(responses_df)
)


# ================================================================
# 6. SEMANTIC SIMILARITY
# ================================================================

print(
    "\nCalculating semantic similarity..."
)

all_responses = responses_df[
    "Response"
].tolist()


embeddings = embedding_model.encode(
    all_responses,
    show_progress_bar=True
)


responses_df[
    "Embedding_Index"
] = range(
    len(responses_df)
)


# ================================================================
# 7. CALCULATE METRICS
# ================================================================

analysis_records = []


for model in models:

    for q in questions:

        qid = q["id"]

        subset = responses_df[
            (responses_df["Model"] == model) &
            (responses_df["Question_ID"] == qid)
        ].sort_values(
            "Day"
        )


        # canonical embedding
        canonical_embedding = embedding_model.encode(
            [q["canonical"]]
        )[0]


        semantic_scores = []


        for _, row in subset.iterrows():

            idx = row[
                "Embedding_Index"
            ]

            response_embedding = embeddings[
                idx
            ]

            similarity = cosine_similarity(
                [canonical_embedding],
                [response_embedding]
            )[0][0]

            semantic_scores.append(
                similarity
            )


        # Pairwise temporal drift
        pairwise_drift = []

        for i in range(
            1,
            len(subset)
        ):

            idx_previous = subset.iloc[
                i - 1
            ]["Embedding_Index"]

            idx_current = subset.iloc[
                i
            ]["Embedding_Index"]


            similarity = cosine_similarity(
                [embeddings[idx_previous]],
                [embeddings[idx_current]]
            )[0][0]


            pairwise_drift.append(
                1 - similarity
            )


        # Average semantic stability
        semantic_stability = (
            np.mean(semantic_scores)
            * 100
        )


        # Mutation count
        mutation_count = (
            subset["Mutation_Type"]
            != "stable"
        ).sum()


        mutation_rate = (
            mutation_count /
            len(subset)
        ) * 100


        contradiction_count = (
            subset["Mutation_Type"]
            == "contradiction"
        ).sum()


        contradiction_rate = (
            contradiction_count /
            len(subset)
        ) * 100


        fact_mutations = (
            subset["Mutation_Type"]
            == "fact_mutation"
        ).sum()


        entity_mutations = (
            subset["Mutation_Type"]
            == "entity_mutation"
        ).sum()


        analysis_records.append({

            "Model":
                model,

            "Question_ID":
                qid,

            "Topic":
                q["topic"],

            "Semantic_Stability":
                semantic_stability,

            "Mutation_Rate":
                mutation_rate,

            "Contradiction_Rate":
                contradiction_rate,

            "Fact_Mutations":
                fact_mutations,

            "Entity_Mutations":
                entity_mutations,

            "Average_Drift":
                np.mean(pairwise_drift)
                if pairwise_drift
                else 0

        })


analysis_df = pd.DataFrame(
    analysis_records
)


# ================================================================
# 8. FACTUAL STABILITY
# ================================================================

def calculate_fact_stability(
    question_data,
    response
):

    response_lower = response.lower()

    detected = 0

    for fact in question_data[
        "facts"
    ]:

        if fact.lower() in response_lower:

            detected += 1


    return (
        detected /
        len(question_data["facts"])
    ) * 100


fact_scores = []


for _, row in responses_df.iterrows():

    q = next(
        x for x in questions
        if x["id"] ==
        row["Question_ID"]
    )

    score = calculate_fact_stability(
        q,
        row["Response"]
    )

    fact_scores.append(
        score
    )


responses_df[
    "Factual_Stability"
] = fact_scores


# ================================================================
# 9. ENTITY STABILITY
# ================================================================

def calculate_entity_stability(
    question_data,
    response
):

    response_lower = response.lower()

    detected = 0

    for entity in question_data[
        "entities"
    ]:

        if entity.lower() in response_lower:

            detected += 1


    return (
        detected /
        len(question_data["entities"])
    ) * 100


entity_scores = []


for _, row in responses_df.iterrows():

    q = next(
        x for x in questions
        if x["id"] ==
        row["Question_ID"]
    )

    score = calculate_entity_stability(
        q,
        row["Response"]
    )

    entity_scores.append(
        score
    )


responses_df[
    "Entity_Stability"
] = entity_scores


# ================================================================
# 10. ECHO STABILITY INDEX
# ================================================================

aggregated = analysis_df.groupby(
    "Model"
).agg({

    "Semantic_Stability":
        "mean",

    "Mutation_Rate":
        "mean",

    "Contradiction_Rate":
        "mean",

    "Average_Drift":
        "mean"

}).reset_index()


for model in models:

    model_rows = responses_df[
        responses_df["Model"] == model
    ]

    factual_stability = (
        model_rows[
            "Factual_Stability"
        ].mean()
    )

    entity_stability = (
        model_rows[
            "Entity_Stability"
        ].mean()
    )


    semantic_stability = (
        aggregated[
            aggregated["Model"] ==
            model
        ]["Semantic_Stability"]
        .iloc[0]
    )


    contradiction_rate = (
        aggregated[
            aggregated["Model"] ==
            model
        ]["Contradiction_Rate"]
        .iloc[0]
    )


    ESI = (

        0.35 *
        semantic_stability

        +

        0.30 *
        factual_stability

        +

        0.20 *
        entity_stability

        +

        0.15 *
        (
            100 -
            contradiction_rate
        )

    )


    aggregated.loc[
        aggregated["Model"] == model,
        "Factual_Stability"
    ] = factual_stability


    aggregated.loc[
        aggregated["Model"] == model,
        "Entity_Stability"
    ] = entity_stability


    aggregated.loc[
        aggregated["Model"] == model,
        "Echo_Stability_Index"
    ] = ESI


# ================================================================
# 11. DISPLAY MAIN RESULTS
# ================================================================

print("\n" + "=" * 90)
print("                    ECHOTRACE RESULTS")
print("=" * 90)

display(
    aggregated.round(2)
)


# ================================================================
# 12. VISUALIZATION — MODEL STABILITY
# ================================================================

plt.figure(
    figsize=(12, 6)
)

sns.barplot(
    data=aggregated,
    x="Model",
    y="Echo_Stability_Index"
)

plt.ylim(
    0,
    100
)

plt.title(
    "EchoTrace — AI Stability Comparison",
    fontsize=18,
    fontweight="bold"
)

plt.ylabel(
    "Echo Stability Index"
)

plt.xlabel(
    "AI Model"
)

plt.tight_layout()

plt.show()


# ================================================================
# 13. VISUALIZATION — SEMANTIC STABILITY
# ================================================================

plt.figure(
    figsize=(12, 6)
)

sns.barplot(
    data=aggregated,
    x="Model",
    y="Semantic_Stability"
)

plt.ylim(
    0,
    100
)

plt.title(
    "Semantic Stability Across AI Models",
    fontsize=18,
    fontweight="bold"
)

plt.ylabel(
    "Semantic Stability (%)"
)

plt.tight_layout()

plt.show()


# ================================================================
# 14. VISUALIZATION — MUTATION RATE
# ================================================================

plt.figure(
    figsize=(12, 6)
)

sns.barplot(
    data=aggregated,
    x="Model",
    y="Mutation_Rate"
)

plt.title(
    "AI Response Mutation Rate",
    fontsize=18,
    fontweight="bold"
)

plt.ylabel(
    "Mutation Rate (%)"
)

plt.tight_layout()

plt.show()


# ================================================================
# 15. VISUALIZATION — CONTRADICTION RATE
# ================================================================

plt.figure(
    figsize=(12, 6)
)

sns.barplot(
    data=aggregated,
    x="Model",
    y="Contradiction_Rate"
)

plt.title(
    "AI Contradiction Rate",
    fontsize=18,
    fontweight="bold"
)

plt.ylabel(
    "Contradiction Rate (%)"
)

plt.tight_layout()

plt.show()


# ================================================================
# 16. TIME-BASED SEMANTIC DRIFT
# ================================================================

timeline_records = []


for model in models:

    for day in days:

        subset = responses_df[
            (responses_df["Model"] == model) &
            (responses_df["Day"] == day)
        ]


        stability = (
            subset[
                "Factual_Stability"
            ].mean()
        )


        semantic = []


        for _, row in subset.iterrows():

            q = next(
                x for x in questions
                if x["id"] ==
                row["Question_ID"]
            )


            canonical_embedding = (
                embedding_model.encode(
                    [q["canonical"]]
                )[0]
            )


            response_embedding = embeddings[
                row["Embedding_Index"]
            ]


            similarity = cosine_similarity(
                [canonical_embedding],
                [response_embedding]
            )[0][0]


            semantic.append(
                similarity * 100
            )


        timeline_records.append({

            "Model":
                model,

            "Day":
                day,

            "Semantic_Stability":
                np.mean(semantic),

            "Factual_Stability":
                stability

        })


timeline_df = pd.DataFrame(
    timeline_records
)


# ================================================================
# 17. VISUALIZATION — DRIFT OVER TIME
# ================================================================

plt.figure(
    figsize=(14, 7)
)

sns.lineplot(
    data=timeline_df,
    x="Day",
    y="Semantic_Stability",
    hue="Model",
    marker="o",
    linewidth=3
)

plt.title(
    "EchoTrace — Semantic Stability Over Time",
    fontsize=18,
    fontweight="bold"
)

plt.ylabel(
    "Semantic Stability (%)"
)

plt.xlabel(
    "Experiment Day"
)

plt.tight_layout()

plt.show()


# ================================================================
# 18. FACTUAL STABILITY OVER TIME
# ================================================================

plt.figure(
    figsize=(14, 7)
)

sns.lineplot(
    data=timeline_df,
    x="Day",
    y="Factual_Stability",
    hue="Model",
    marker="o",
    linewidth=3
)

plt.title(
    "EchoTrace — Factual Stability Over Time",
    fontsize=18,
    fontweight="bold"
)

plt.ylabel(
    "Factual Stability (%)"
)

plt.xlabel(
    "Experiment Day"
)

plt.tight_layout()

plt.show()


# ================================================================
# 19. MUTATION TYPE ANALYSIS
# ================================================================

mutation_counts = (
    responses_df
    .groupby(
        ["Model", "Mutation_Type"]
    )
    .size()
    .reset_index(
        name="Count"
    )
)


plt.figure(
    figsize=(14, 7)
)

sns.barplot(
    data=mutation_counts,
    x="Model",
    y="Count",
    hue="Mutation_Type"
)

plt.title(
    "EchoTrace — Mutation Type Distribution",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()

plt.show()


# ================================================================
# 20. HEATMAP — QUESTION STABILITY
# ================================================================

question_heatmap = analysis_df.pivot(
    index="Question_ID",
    columns="Model",
    values="Semantic_Stability"
)


plt.figure(
    figsize=(10, 8)
)

sns.heatmap(
    question_heatmap,
    annot=True,
    fmt=".1f",
    cmap="RdYlGn",
    vmin=0,
    vmax=100
)

plt.title(
    "EchoTrace — Question Stability Heatmap",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "AI Model"
)

plt.ylabel(
    "Question"
)

plt.tight_layout()

plt.show()


# ================================================================
# 21. DRIFT VS STABILITY SCATTER
# ================================================================

plt.figure(
    figsize=(12, 7)
)

sns.scatterplot(
    data=analysis_df,
    x="Average_Drift",
    y="Semantic_Stability",
    hue="Model",
    size="Contradiction_Rate",
    sizes=(50, 300)
)

plt.title(
    "EchoTrace — Drift vs Semantic Stability",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "Average Semantic Drift"
)

plt.ylabel(
    "Semantic Stability"
)

plt.tight_layout()

plt.show()


# ================================================================
# 22. RESPONSE MUTATION NETWORK
# ================================================================

print(
    "\nBuilding response mutation network..."
)

G = nx.Graph()


selected_question = "Q001"

selected = responses_df[
    responses_df["Question_ID"] ==
    selected_question
]


for _, row in selected.iterrows():

    node = (
        row["Model"]
        + "_D"
        + str(row["Day"])
    )

    G.add_node(
        node,
        model=row["Model"],
        day=row["Day"],
        mutation=row["Mutation_Type"]
    )


nodes = list(
    G.nodes()
)


for i in range(
    len(nodes)
):

    for j in range(
        i + 1,
        len(nodes)
    ):

        node_a = nodes[i]
        node_b = nodes[j]

        idx_a = responses_df[
            (
                responses_df["Model"]
                + "_D"
                + responses_df["Day"].astype(str)
            ) == node_a
        ]

        idx_b = responses_df[
            (
                responses_df["Model"]
                + "_D"
                + responses_df["Day"].astype(str)
            ) == node_b
        ]

        if len(idx_a) == 0 or len(idx_b) == 0:
            continue


# ================================================================
# 23. RESPONSE COMPARISON EXAMPLE
# ================================================================

print("\n" + "=" * 90)
print("             RESPONSE MUTATION EXAMPLE")
print("=" * 90)


example_question = questions[0]


example_responses = responses_df[
    responses_df["Question_ID"] ==
    example_question["id"]
].sort_values(
    ["Model", "Day"]
)


for model in models:

    print(
        "\n" +
        "─" * 80
    )

    print(
        "MODEL:",
        model
    )

    model_data = example_responses[
        example_responses["Model"] ==
        model
    ]


    for _, row in model_data.iterrows():

        print(
            f"\nDay {row['Day']} "
            f"[{row['Mutation_Type']}]"
        )

        print(
            row["Response"]
        )


# ================================================================
# 24. FIND MOST UNSTABLE QUESTIONS
# ================================================================

question_ranking = (
    analysis_df
    .groupby(
        ["Question_ID", "Topic"]
    )
    .agg({

        "Semantic_Stability":
            "mean",

        "Mutation_Rate":
            "mean",

        "Contradiction_Rate":
            "mean",

        "Average_Drift":
            "mean"

    })
    .reset_index()
)


question_ranking[
    "Instability_Score"
] = (

    0.45 *
    (100 -
     question_ranking[
         "Semantic_Stability"
     ])

    +

    0.30 *
    question_ranking[
        "Mutation_Rate"
    ]

    +

    0.25 *
    question_ranking[
        "Contradiction_Rate"
    ]

)


question_ranking = question_ranking.sort_values(
    "Instability_Score",
    ascending=False
)


print("\n" + "=" * 90)
print("               MOST UNSTABLE QUESTIONS")
print("=" * 90)

display(
    question_ranking.round(2)
)


# ================================================================
# 25. MODEL RELIABILITY REPORT
# ================================================================

print("\n" + "=" * 90)
print("                  ECHOTRACE REPORT")
print("=" * 90)


for _, row in aggregated.iterrows():

    model = row["Model"]

    score = row[
        "Echo_Stability_Index"
    ]

    if score >= 90:

        status = "HIGH STABILITY"

    elif score >= 75:

        status = "MODERATE STABILITY"

    else:

        status = "HIGH MUTATION RISK"


    print(
        f"""
MODEL: {model}

Echo Stability Index : {score:.2f}

Semantic Stability   : {row['Semantic_Stability']:.2f}%
Factual Stability    : {row['Factual_Stability']:.2f}%
Entity Stability     : {row['Entity_Stability']:.2f}%
Mutation Rate        : {row['Mutation_Rate']:.2f}%
Contradiction Rate   : {row['Contradiction_Rate']:.2f}%

STATUS               : {status}

"""
    )


# ================================================================
# 26. EXPERIMENTAL FINDINGS
# ================================================================

print("\n" + "=" * 90)
print("                  KEY FINDINGS")
print("=" * 90)


best_model = aggregated.loc[
    aggregated[
        "Echo_Stability_Index"
    ].idxmax()
]


worst_model = aggregated.loc[
    aggregated[
        "Echo_Stability_Index"
    ].idxmin()
]


most_unstable_question = (
    question_ranking.iloc[0]
)


print(
    f"""
1. Most stable simulated model:
   {best_model['Model']}

2. Lowest stability simulated model:
   {worst_model['Model']}

3. Most unstable question:
   {most_unstable_question['Question_ID']}

4. Average semantic stability:
   {aggregated['Semantic_Stability'].mean():.2f}%

5. Average mutation rate:
   {aggregated['Mutation_Rate'].mean():.2f}%

6. Average contradiction rate:
   {aggregated['Contradiction_Rate'].mean():.2f}%

7. Questions monitored:
   {len(questions)}

8. Models monitored:
   {len(models)}

9. Experiment time points:
   {len(days)}

10. Total AI responses analyzed:
    {len(responses_df)}
"""
)


# ================================================================
# 27. IMPORTANT RESEARCH LIMITATION
# ================================================================

print("\n" + "=" * 90)
print("                  RESEARCH LIMITATION")
print("=" * 90)

print(
"""
This notebook uses a controlled response simulator to demonstrate
the EchoTrace methodology without requiring an external LLM API.

Therefore:

• The measured mutation rates are NOT measurements of real LLMs.
• The Echo Stability Index is an experimental metric.
• The simulated models do not represent actual commercial models.
• A real research study should collect repeated responses from
  real models under controlled experimental conditions.
• Temperature, system prompts, model versions, context windows,
  retrieval sources and sampling seeds should be recorded.
"""
)


# ================================================================
# 28. EXPORT ALL RESULTS
# ================================================================

responses_df.to_csv(
    "echotrace_responses.csv",
    index=False
)

analysis_df.to_csv(
    "echotrace_question_analysis.csv",
    index=False
)

aggregated.to_csv(
    "echotrace_model_scores.csv",
    index=False
)

timeline_df.to_csv(
    "echotrace_timeline.csv",
    index=False
)

question_ranking.to_csv(
    "echotrace_question_ranking.csv",
    index=False
)

mutation_counts.to_csv(
    "echotrace_mutation_counts.csv",
    index=False
)


# ================================================================
# 29. FINAL DASHBOARD
# ================================================================

print("\n" + "=" * 90)
print("                    FINAL ECHOTRACE DASHBOARD")
print("=" * 90)


fig, axes = plt.subplots(
    2,
    2,
    figsize=(18, 12)
)


# ---- Plot 1 ----

sns.barplot(
    data=aggregated,
    x="Model",
    y="Echo_Stability_Index",
    ax=axes[0,0]
)

axes[0,0].set_title(
    "Echo Stability Index",
    fontsize=15,
    fontweight="bold"
)

axes[0,0].set_ylim(
    0,
    100
)


# ---- Plot 2 ----

sns.lineplot(
    data=timeline_df,
    x="Day",
    y="Semantic_Stability",
    hue="Model",
    marker="o",
    ax=axes[0,1]
)

axes[0,1].set_title(
    "Semantic Stability Over Time",
    fontsize=15,
    fontweight="bold"
)


# ---- Plot 3 ----

sns.barplot(
    data=aggregated,
    x="Model",
    y="Mutation_Rate",
    ax=axes[1,0]
)

axes[1,0].set_title(
    "Mutation Rate",
    fontsize=15,
    fontweight="bold"
)


# ---- Plot 4 ----

sns.barplot(
    data=aggregated,
    x="Model",
    y="Contradiction_Rate",
    ax=axes[1,1]
)

axes[1,1].set_title(
    "Contradiction Rate",
    fontsize=15,
    fontweight="bold"
)


plt.suptitle(
    "ECHOTRACE AI — AI MEMORY & RELIABILITY OBSERVATORY",
    fontsize=22,
    fontweight="bold"
)

plt.tight_layout(
    rect=[0, 0, 1, 0.95]
)

plt.show()


# ================================================================
# 30. COMPLETION
# ================================================================

print("\n")

print("=" * 90)

print(
"""
                  ECHOTRACE AI COMPLETE

Your research prototype contains:

✓ Longitudinal AI response tracking
✓ Semantic drift detection
✓ Factual stability analysis
✓ Entity stability analysis
✓ Contradiction detection
✓ Mutation classification
✓ Model comparison
✓ Time-based analysis
✓ Question stability ranking
✓ Echo Stability Index
✓ Multiple research visualizations
✓ AI reliability report
✓ CSV result exports

Files generated:

• echotrace_responses.csv
• echotrace_question_analysis.csv
• echotrace_model_scores.csv
• echotrace_timeline.csv
• echotrace_question_ranking.csv
• echotrace_mutation_counts.csv

PROJECT STATUS:
RESEARCH PROTOTYPE COMPLETE
"""
)

print("=" * 90)
```
